OUTLIERS

```python
print("CHECKING WITH FULL DETAIL")
print()

numerical_columns = data.select_dtypes(include="number").columns

for col in numerical_columns:
  Q1 = data[col].quantile(0.25)
  Q3 = data[col].quantile(0.75)

  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]

  print(f"Column: {col}")
  print()
  print(f"Lower Bound: {lower_bound:.2f}")
  print(f"Upper Bound: {upper_bound:.2f}")
  print(f"Number of Outliers %: {len(outliers) / len(data) * 100 :.2f}")
  print("---" * 40)
```

Remove outlers

```python
outlier_columns = [
    "age",
    "weight",
    "bmi",
    "bloodpressure",
]

Q1 = data[outlier_columns].quantile(0.25)
Q3 = data[outlier_columns].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

keep = (
    (data[outlier_columns] >= lower_bound) &
    (data[outlier_columns] <= upper_bound)
).all(axis=1)

print("Before:", len(data))

data_clean = data[keep].copy()

print("After:", len(data_clean))
print("Removed:", len(data) - len(data_clean))
```

X , y (independent variable , depended variable)

```python
X = data.drop('claim', axis=1)
y = data['claim']
```

TRAIN_TEST_SPLIT

```python
from  sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(X ,y , test_size=0.20 , random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer


categorical_columns = x_train.select_dtypes(
    include=["object", "category"]
).columns

numerical_columns = x_train.select_dtypes(
    include=["int64", "float64"]
).columns


print("\nCategorical columns:")
print(list(categorical_columns))

print("\nNumerical columns:")
print(list(numerical_columns))



#  CATEGORICAL PIPELINE


cat_pipeline = Pipeline([
    ("imputer" ,SimpleImputer(strategy="most_frequent")),
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])



#  NUMERICAL PIPELINE


num_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="mean")),
    ("scaler",StandardScaler())
])



#  COLUMN TRANSFORMER


preprocessor = ColumnTransformer([
    ("categorical",cat_pipeline,categorical_columns),
    ("numerical",num_pipeline,numerical_columns)
])



#  FIT TRAIN + TRANSFORM TRAIN


x_train_processed = preprocessor.fit_transform(x_train)



#  TRANSFORM TEST


x_test_processed = preprocessor.transform(
    x_test
)

#  CHECK


print("\nOriginal train shape:")
print(x_train.shape)

print("\nProcessed train shape:")
print(x_train_processed.shape)

print("\nOriginal test shape:")
print(x_test.shape)

print("\nProcessed test shape:")
print(x_test_processed.shape)
```

FEATURE SELECTION 

```python
from sklearn.feature_selection import SelectKBest, r_regression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor
import numpy as np
import pandas as pd

k_values = [5, 10, 20, 30, 40, 50, 75, 100]
k_values = [k for k in k_values if k <= x_train_processed.shape[1]]

results = []

for k in k_values:

    selector = SelectKBest(score_func=r_regression,k=k)

    x_train_selected = selector.fit_transform(x_train_processed,y_train)

    x_test_selected = selector.transform(x_test_processed)

    model = XGBRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=6,
        random_state=42
    )

    model.fit(x_train_selected, y_train)

    y_pred = model.predict(x_test_selected)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    results.append({
        "mae":mae , 
        "rmse" : rmse , 
        "r2": r2
    })


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="r2",
    ascending=False
)

print(results_df)
```

RANDOMIZED SEARCH

```python
from sklearn.model_selection import GridSearchCV
import warnings
from sklearn.model_selection import KFold
warnings.filterwarnings("ignore")

from sklearn.model_selection import RandomizedSearchCV

best_models = {}

for name, model in models.items():

    hyper = RandomizedSearchCV(
        estimator=model,
        param_distributions=params[name],
        n_iter=10,          
        cv=3,               
        scoring="neg_mean_absolute_error",
        n_jobs=-1,
        random_state=42
    )
    
    hyper.fit(x_train_selected, y_train)
    
    best_models[name] = hyper.best_estimator_
    
    print("\n", name)
    print("Best Parameters:", hyper.best_params_)
    print("Best CV neg mean squared error:", -hyper.best_score_)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
import numpy as np
import pandas as pd

results = []

for name, model in best_models.items():

    # Predict on untouched test data
    y_pred = model.predict(x_test_selected)

    # Metrics
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)

    results.append({
        "Model": name,
        "R2": r2,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })

results_df = pd.DataFrame(results)

# Sort by R2
results_df = results_df.sort_values(by="R2",ascending=False)

print(results_df)
```
